```python

# Module 1 - Operators and Conditions

a = 10
b = 5

# Arithmetic operators
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

# Comparison operators
print("Is a greater than b?", a > b)
print("Are a and b equal?", a == b)

# If-else condition
marks = 75

if marks >= 40:
    print("Result: Pass")
else:
    print("Result: Fail")

# If-elif-else condition
if marks >= 80:
    print("Grade: A")
elif marks >= 60:
    print("Grade: B")
else:
    print("Grade: C")
```

```python

# Module 1 - Introduction to Data Science

# Sample student dataset
students = ["Asha", "Rahul", "Priya", "Kiran", "Anil"]
marks = [85, 70, 92, 60, 78]

# Display student data
print("Student Names:", students)
print("Student Marks:", marks)

# Calculate average marks
average = sum(marks) / len(marks)
print("Average Marks:", average)

# Find highest and lowest marks
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))

# Identify students who scored 75 or more
print("Students scoring 75 or more:")
for i in range(len(students)):
    if marks[i] >= 75:
        print(students[i], "-", marks[i])
```


## Data Science Concepts and Real-World Applications

### What is Data Science?
Data Science is the process of collecting, organizing, analyzing, and interpreting data to discover useful information and support decision-making.

### Basic Data Science Steps
1. Data Collection: Gathering data from different sources.
2. Data Cleaning: Fixing missing or incorrect data.
3. Data Analysis: Finding patterns and calculating useful statistics.
4. Data Interpretation: Understanding what the results mean.
5. Decision-Making: Using the results to make better decisions.

### Real-World Applications
- Education: Analyzing student marks and academic performance.
- Healthcare: Studying health data to support medical decisions.
- Business: Understanding customer preferences and sales trends.
- Banking: Identifying unusual transactions and assessing financial risks.
- Transportation: Analyzing traffic and travel patterns.

### Practical Example
In this module, student marks were analyzed using Python. The program calculated average, highest, and lowest marks and identified students scoring 75 or more. This demonstrates how data analysis can help understand student performance.
